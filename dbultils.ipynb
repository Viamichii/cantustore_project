```python
files = dbutils.fs.ls(RAW)

for f in files:
    name = f.name
    
    if name.endswith(".parquet"):
        folder = name.replace(".parquet","")
        dbutils.fs.mkdirs(f"{RAW}/{folder}")
        
    elif name.endswith(".csv"):
        folder = name.replace(".csv","")
        dbutils.fs.mkdirs(f"{RAW}/{folder}")
```

```python
files = dbutils.fs.ls(RAW)

for f in files:
    name = f.name
    src = f.path

    if name.endswith(".parquet"):
        folder = name.replace(".parquet","")
        dest = f"{RAW}/{folder}/{name}"
        dbutils.fs.mv(src, dest)

    elif name.endswith(".csv"):
        folder = name.replace(".csv","")
        dest = f"{RAW}/{folder}/{name}"
        dbutils.fs.mv(src, dest)
```

```python
RAW = "dbfs:/Volumes/workspace/cantustore/raw_prova_dados"

folders = [f.name.rstrip("/") for f in dbutils.fs.ls(RAW) if f.isDir()]

for t in folders:
    path = f"{RAW}/{t}"
    files = dbutils.fs.ls(path)

    print("\n==============================")
    print("RAW tabela:", t)
    print("==============================")

    if any(f.name.endswith(".parquet") for f in files):
        df = spark.read.parquet(f"{path}/*.parquet")

    elif any(f.name.endswith(".csv") for f in files):
        df = (spark.read
              .option("header", True)
              .option("inferSchema", True)
              .option("sep", "|")
              .option("quote", '"')
              .csv(f"{path}/*.csv"))
    else:
        print("Nenhum parquet/csv encontrado")
        continue

    print("Colunas:", df.columns)
    display(df.limit(200))
```
